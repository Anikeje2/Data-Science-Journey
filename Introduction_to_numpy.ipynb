```python
df = pd.read_csv('/content/drive/MyDrive/SCA class/students_messy_data.csv')
df.head() #to load messy data onto the notebook
```

```python
print("\n Percentage missing for each column \n") #finding missing data per column
print(df.isnull().sum()/df.shape[0]*100)
```

```python
# handling age missing data
mean_age = df['Age'].mean()
df['Age'].fillna(mean_age, inplace=True)
df.head()
```

```python
#handling missing scores
english_score = pd.to_numeric(df["English"], errors="coerce").median()
df["English"].fillna(english_score, inplace=True)
```

```python
#handling missing scores
math_score = pd.to_numeric(df["Math"], errors="coerce").median()
df["Math"].fillna(math_score, inplace=True)
```

```python
#handling missing scores
sci_score = pd.to_numeric(df["Science"], errors="coerce").median()
df["Science"].fillna(sci_score, inplace=True)
```

```python
#handling missing GPA
median_gpa = pd.to_numeric(df["GPA"], errors="coerce").median()
df["GPA"].fillna(median_gpa, inplace=True)
```

```python
#handling missing Gender nd city
df["Gender"] = df["Gender"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")
```

```python
df['EnrollmentDate'].fillna(pd.to_datetime('02-03-2026'), inplace=True)
 #replacing all missing data with todays date
df.head()
```

```python
print(df.isnull().sum()/df.shape[0]*100)
```

```python
#converting data types
print(df['StudentID'].dtype)
print(df['FirstName'].dtype)
print(df['LastName'].dtype)
print(df['Age'].dtype)
print(df['Gender'].dtype)
print(df['Email'].dtype)
print(df['City'].dtype)
print(df['EnrollmentDate'].dtype)
print(df['Department'].dtype)
print(df['GPA'].dtype)
print(df['Math'].dtype)
print(df['English'].dtype)
print(df['Science'].dtype)
print(df['Status'].dtype)
print(df['ScholarshipAmount'].dtype)
```

```python
#converting data type
df['StudentID'] = df['StudentID'].astype(str)
print(df['StudentID'].dtype)

df['Age'] = df['Age'].astype(int)
print(df['Age'].dtype)

```

```python
#standardizing text
df['Gender']= df['Gender'].replace('male','M')
df['Gender']= df['Gender'].replace('female','F')
df['Gender']= df['Gender'].replace('Male','M')
df['Gender']= df['Gender'].replace('Female','F')
df.head(5)
```

```python
df['FirstName'] = df['FirstName'].str.strip() #removing whitespaces from all strings
df['LastName'] = df['LastName'].str.strip()
df['Gender'] = df['Gender'].str.strip()
df['Email'] = df['Email'].str.strip()
df['City'] = df['City'].str.strip()
df['Department'] = df['Department'].str.strip()
df['Status'] = df['Status'].str.strip()
```

```python
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"], errors="coerce")
print(df["EnrollmentDate"].dtype)
```

```python
df['Total'] = df['English']+df['Math']+df['Science']  #finding sum of all scores and including Total column
df.head()
```

```python
df['Avg_scores'] =  df['Total']/3    #finding and adding average columns
df.head()
```

```python
active_std= df[df['Status']=='Active'] #filtering all active students
print(active_std)
```

```python
mean_gpa = df['GPA'].mean() #calculating mean GPA
print(mean_gpa)
```

```python
mean_total =df['Total'].mean()
print(mean_total)
```

```python
print(df.describe())
```

```python
conditions = [df['English'] >= 70, (df['English'] < 70) & (df['English'] >= 55), (df['English'] < 55) & (df['English'] >= 45)]
choices = ['A', 'B', 'C']
df['EnglishGrade'] = np.select(conditions, choices, default='D')
df.head() #Assigning grades to scores 70 and above is A, and less 70 to 55 is B, C is less than 55 to 45 and others is D
```

```python
conditions = [df['Math'] >= 70, (df['Math'] < 70) & (df['Math'] >= 55), (df['Math'] < 55) & (df['Math'] >= 45)]
choices = ['A', 'B', 'C']
df['MathGrade'] = np.select(conditions, choices, default='D')
df.head()
```

```python
conditions = [df['Science'] >= 70, (df['Science'] < 70) & (df['Science'] >= 55), (df['Science'] < 55) & (df['Science'] >= 45)]
choices = ['A', 'B', 'C']
df['ScienceGrade'] = np.select(conditions, choices, default='D')
df.head()
```

Documentation
 To clean and prepare the dataset, I first assessed missing values across all columns to understand where gaps existed. For numerical data, I used the mean for Age and the median for scores and CGPA to preserve distribution integrity. Categorical fields like Gender and City were filled with "Unknown" to avoid assumptions, while all string columns were stripped of extra spaces and standardized for consistency. Missing Enrollment Dates were replaced with a placeholder to maintain chronological completeness. I also converted all columns to their correct data types and created new features, including Total and Average Scores, alongside a letter grading system (A–D) based on student performance. Finally, the dataset was filtered to include only Active students, ensuring the analysis focuses on the current student body.
