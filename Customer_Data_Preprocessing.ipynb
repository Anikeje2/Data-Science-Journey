```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer  #importing libraries
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
```

```python
df=pd.read_csv('/content/drive/MyDrive/SCA class/customer_data.csv') # loading dataset
```

```python
df.info()
print('------------')
df.describe()
```

```python
df.isnull().sum() #checking missing data
```

```python
#using simple imputer to fill all missing age with the median age
imputer = SimpleImputer(strategy = 'median')
df['Age']= imputer.fit_transform(df[['Age']])


#using simple imputer to fill all missing gender with the mode,the most frequent gender
impute =SimpleImputer(strategy='most_frequent')
df['Gender']= impute.fit_transform(df[['Gender']])[:, 0]
```

Median was used for Age to reduce the effect of outliers. Mode was used for Gender because it is a categorical data.

```python
df = df.dropna(subset=['Income']) #dropping missing income
```

```python
df.isnull().sum() #rechecking missing data
```

```python
#converting gender to binary by label encoding
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
```

Gender was label encoded to convert it into numerical form because it is a ordered data. So it is binary now

```python
df = pd.get_dummies(df, columns=['City'])
```

```python
df['Purchased'] = le.fit_transform(df['Purchased']) #converting purchase to binary by label encoding
```

```python


age = df[['Age']]   #using standard scale on age
scaler = StandardScaler()
age_scaled = scaler.fit_transform(age)
```

```python
df['Age_scaled'] = age_scaled  #comparing original age and scaled age
df[['Age', 'Age_scaled']].head()
```

```python
income =df[['Income']]
scaler = MinMaxScaler()
income_scaled = scaler.fit_transform(income)
```

```python
df['Income_scaled'] = income_scaled #comparing original income and scaled income
df[['Income', 'Income_scaled']].head()
```

The City variable was one-hot encoded to prevent false relationships between categories. Gender and Purchased were label encoded to convert them into numerical form suitable for modeling. Age and Income were scaled using StandardScaler and MinMaxScaler to ensure consistent feature ranges and improve model performance.
