A pharmaceutical company wants to check patient compliance to a new soft jell tablet so they measure number of doses per week
Standard tablet average compliance = 5 doses/week

```python
import scipy.stats as stats #importing libraries
import numpy as np
from scipy.stats import chi2_contingency
```

```python
#Assignment 1
weekly_dose= [6, 7, 5, 6, 8, 7, 6, 7, 5, 6,
7, 8, 6, 7, 5, 6, 7, 8, 6, 7,
5, 6, 7, 6, 8, 7, 6, 7, 5, 6,
7, 8, 6, 7, 5, 6, 7, 8, 6, 7]
```

```python
# H0: mu = 5, H1: mu!=5 my hypotheses
t_stat, p_value = stats.ttest_1samp(
    weekly_dose, popmean=5)

print(f'T-statistic: {t_stat:.4f}') #finding T-statistic
print(f'P-value:     {p_value:.4f}') #finding p-value

if p_value < 0.05:
    print('Reject H0')
else:
    print('Fail to reject H0')
```

Since the p-value < 0.05 is less than the level of significance, the null hypothesis is rejected. This indicates that there is a statistically significant difference between the mean weekly dose taken and the hypothesized value of 5 doses per week.

```python
# Assignment 2
Salaries=[52000, 48000, 61000, 45000, 53000,
55000, 40000, 47000, 60000, 51000,
49000, 58000, 43000, 52000, 54000,
50000, 46000, 62000, 39000, 57000,
52000, 48000, 56000, 44000, 53000]
# H0: mu = 55,000, H1: mu!=55,000 my hypotheses
```

```python
t_stat,p_value = stats.ttest_1samp(Salaries, 55000)
print(f'T-statistic: {t_stat:.4f}') #finding T-statistic
print(f'P-value:     {p_value:.4f}') #finding p-value
if p_value < 0.05:
    print('Reject H0')
else:
    print('Fail to reject H0')
```

The Null hypothesis is rejected because p-value< 0.05 There is sufficient statistical evidence to conclude that the average salary of employees is different from $55,000. In this sample, the average salary is significantly lower than the company’s claimed $55,000.

```python
before = [142, 158, 135, 170, 125, 163, 148, 139, 175, 122, 155, 168, 130, 145, 160]
after  = [160, 172, 148, 185, 140, 178, 165, 150, 190, 138, 170, 182, 147, 162, 175]
```

```python
# H0: no difference   H1: after != before  The hypotheses
t, p = stats.ttest_rel(before, after)
print(f"T-statistic: {t:.4f}")  #finding T-statistic
print(f"P-value:     {p:.4f}")   #finding p-value

if p < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

Reject H0
The Null hypothesis is rejected because p-value < 0.05.
There is sufficient statistical evidence to conclude that
sales after the campaign are significantly different from before.
In this sample, the average sales significantly increased
after the marketing campaign was launched.

```python
factory_a = [88, 92, 85, 90, 87, 93, 89, 84, 91, 86,
             90, 88, 92, 85, 89, 87, 91, 86, 90, 88]

factory_b = [82, 79, 85, 80, 83, 78, 81, 84, 77, 82,
             80, 83, 79, 81, 78, 84, 80, 82, 79, 83]

# H0: mu_A = mu_B    H1: mu_A != mu_B
t_stat, p_value = stats.ttest_ind(factory_a, factory_b)
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value:     {p_value:.4f}")

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

The Null hypothesis is rejected because p-value < 0.05.
There is sufficient statistical evidence to conclude that
the average widget quality of Factory A and Factory B are different.
In this sample, Factory A produces significantly higher
quality widgets than Factory B.

```python
# H0: Sign-up rate is independent of page design
# H1: Sign-up rate is associated with page design

table = [[45, 455],
         [62, 438]]

chi2, p_value_chi, dof, expected = chi2_contingency(table)

print(f"Chi2-statistic: {chi2:.4f}")
print(f"P-value:        {p_value_chi:.4f}")
print(f"Degrees of freedom: {dof}")

if p_value_chi < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

The p-value (0.1017) is greater than 0.05, so we fail to reject the null hypothesis. Therefore, there is insufficient evidence to conclude that page design significantly affects sign-up rates.
